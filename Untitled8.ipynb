# Pet Supplies Dataset Analysis
## Overview
This project performs exploratory data analysis (EDA) and data cleaning on a dataset containing information about pet supplies. The primary steps include inspecting, cleaning, analyzing the data, and visualizing key trends.

### Import required libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
```

### Step 1: Load the dataset

```python
file_path = r'C:\Users\yugan\Downloads\pet_supplies_2212.csv'  # Update with the correct file path
data = pd.read_csv(file_path)
# Display the first few rows
print(data.head())
```

### Step 2: Initial data inspection

```python
print("First 5 rows of the dataset:")
print(data.head())

print("\nDataset Info:")
print(data.info())

print("\nSummary Statistics:")
print(data.describe(include='all'))

```

### Step 3: Data Cleaning

```python
# Convert 'price' to numeric, handling errors by setting invalid entries to NaN
data['price'] = pd.to_numeric(data['price'], errors='coerce')

# Fill missing values in 'price' with the median
data['price'] = data['price'].fillna(data['price'].median())

# Fill missing values in 'rating' with the mean
data['rating'] = data['rating'].fillna(data['rating'].mean())

# Check for any remaining missing values
print("\nMissing Values After Cleaning:")
print(data.isnull().sum())
```

### Step 4: Add New Columns for Analysis

```python
# Add a 'sales_per_unit_price' column
data['sales_per_unit_price'] = data['sales'] / data['price']
# Add a 'is_high_rating' column (1 if rating >= 5, else 0)
data['is_high_rating'] = np.where(data['rating'] >= 5, 1, 0)
```

### Step 5: Export the Cleaned Dataset

```python
# Specify the path where you want to save the file
cleaned_file_path = r'C:\Users\yugan\Downloads\pet_supplies_2212.csv' 

# Ensure the directory exists
output_dir = os.path.dirname(cleaned_file_path)
if not os.path.exists(output_dir):
    os.makedirs(output_dir)  # Create the directory if it doesn't exist

# Save the cleaned dataset
data.to_csv(cleaned_file_path, index=False)
print(f"\nCleaned data saved to {cleaned_file_path}")
```

### Step 6: Exploratory Data Analysis (Visualizations)
#### 6.1: Sales Distribution

```python
plt.figure(figsize=(10, 6))
sns.histplot(data['sales'], bins=30, kde=True, color='blue')
plt.title('Distribution of Sales', fontsize=16)
plt.xlabel('Sales', fontsize=12)
plt.ylabel('Frequency', fontsize=12)
plt.show()
```

#### 6.2: Sales by Category

```python
plt.figure(figsize=(12, 6))

# Assign 'category' to hue for proper color handling and set legend=False if not required
sns.barplot(
    x='category', 
    y='sales', 
    data=data, 
    errorbar=None,  # Updated for the ci deprecation
    hue='category',  # Assign category to hue
    dodge=False,     # Prevents splitting of bars by hue
    palette='viridis'
)

plt.title('Average Sales by Category', fontsize=16)
plt.xlabel('Category', fontsize=12)
plt.ylabel('Average Sales', fontsize=12)
plt.xticks(rotation=45)
plt.legend([], [], frameon=False)  # Optional: Hide legend if not needed
plt.show()

```

#### 6.3: Rating Distribution

```python
plt.figure(figsize=(10, 6))
sns.countplot(
    x='rating', 
    data=data, 
    palette='muted', 
    hue='rating',  # Assign x variable to hue
    dodge=False    # Prevent bar separation for hue
)
plt.title('Distribution of Ratings', fontsize=16)
plt.xlabel('Rating', fontsize=12)
plt.ylabel('Count', fontsize=12)
plt.legend([], [], frameon=False)  # Hide the legend
plt.show()
```

#### 6.4: Price vs. Sales Scatter Plot

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(x='price', y='sales', hue='category', data=data, palette='deep')
plt.title('Price vs. Sales', fontsize=16)
plt.xlabel('Price', fontsize=12)
plt.ylabel('Sales', fontsize=12)
plt.legend(title='Category', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.show()
```

#### 6.5: Repeat Purchases Analysis

```python
repeat_purchase_counts = data['repeat_purchase'].value_counts()

plt.figure(figsize=(8, 6))
sns.barplot(
    x=repeat_purchase_counts.index,
    y=repeat_purchase_counts.values,
    palette='pastel',
    hue=repeat_purchase_counts.index,  # Assign x variable to hue
    dodge=False  # Avoid separating bars
)
plt.title('Repeat Purchase Analysis', fontsize=16)
plt.xlabel('Repeat Purchase (0 = No, 1 = Yes)', fontsize=12)
plt.ylabel('Count', fontsize=12)
plt.legend([], [], frameon=False)  # Hide legend if not needed
plt.show()
```

#### 6.6: Correlation Heatmap

```python
plt.figure(figsize=(12, 8))

# Compute correlation matrix for numeric columns only
numeric_data = data.select_dtypes(include=[np.number])
correlation_matrix = numeric_data.corr()

# Plot heatmap
sns.heatmap(
    correlation_matrix, 
    annot=True, 
    cmap='coolwarm', 
    fmt='.2f', 
    linewidths=0.5  # Add gridlines for better readability
)

plt.title('Correlation Heatmap', fontsize=16)
plt.show()
```
